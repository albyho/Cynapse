# Scroll alignment fixture

`scroll_align.ipynb` is long on purpose: `align_target` sits far below the
first viewport and `align_tall` is taller than the viewport, so every
`JupyterLabCellSelect.align` value has a visible, distinct result.

- `start`: target top at the viewport top
- `center`: target middle at the viewport middle
- `end`: target bottom at the viewport bottom
- `auto`: minimal scroll; no movement when the target is already visible
- `top-center`: tall Cell top at the viewport middle

Each value is driven by its own Step; advance the Step to observe it.
